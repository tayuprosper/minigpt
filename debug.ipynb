```python
import numpy as np
from llm.model import load_model
from llm.tokenizer import CharTokenizer

model, chars = load_model("checkpoints/fables.npz")
tok = CharTokenizer(chars)
text = open("data/fables.txt", encoding="utf-8").read()
data = np.array(tok.encode(text))

x = data[:64][None, :]        # one window of 64 characters, shape (1, 64)
y = data[1:65][None, :]       # the same window shifted by one
print(float(model(x, y)[1].data))
```
