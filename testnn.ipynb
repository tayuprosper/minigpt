```python
import numpy as np
from llm.tensor import Tensor, numerical_grad
from llm.nn import softmax, cross_entropy


rng = np.random.default_rng(0)
z = Tensor(rng.normal(size=(5, 4)))

targets = np.array([0, 3, 1, 1, 2])
w = rng.normal(size=(5,4))

cross_entropy(z, targets).backward()

f = lambda: float(cross_entropy(Tensor(z.data), targets).data)
print("cross_entropy", np.abs(z.grad - numerical_grad(f, z.data)).max())
z.grad = None
(softmax(z) * Tensor(w)).sum().backward()
f = lambda: float((softmax(Tensor(z.data)).data * w).sum())
print("softmax", np.abs(z.grad - numerical_grad(f, z.data)).max())
           
```
