```python
import numpy as np
from llm.tensor import Tensor
from llm.nn import Dropout, manual_seed

manual_seed(0)


drop = Dropout(0.5)
x = Tensor(np.ones((2, 8), dtype=np.float32))
print(drop(x).data)
drop.eval()

print(drop(x).data)
```

```python
rng = np.random.default_rng(0)

x_plain = x_norm = rng.normal(size=(8,32))

for layer in range(20):
    W = rng.normal(0, 0.3, size=(32,32))
    x_plain = x_plain @ W
    x_norm = drop(x_norm @ W)
    m = x_norm.mean(axis=0, keepdims=True)
    s = x_norm.std(axis=0, keepdims=True)
    x_norm = (x_norm - m) / s
print(f"without: {x_plain.std():.2e} with: {x_norm.std():.2e}")
```
