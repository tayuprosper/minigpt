```python
import numpy as np
from llm.tensor import Tensor, unbroadcast

# Check 1: unbroadcast should keep the shape when it already matches
print(unbroadcast(np.ones((2, 4, 6)), (2, 4, 6)).shape)   # expect (2, 4, 6)
print(unbroadcast(np.ones((2, 4, 6)), (6,)).shape)        # expect (6,)

# Check 2: the simplest matmul backward
a = Tensor(np.ones((2, 3)))
b = Tensor(np.ones((3, 4)))
(a @ b).sum().backward()
print(a.grad.shape, b.grad.shape)                         # expect (2, 3) (3, 4)

# Check 3: add with a bias
a = Tensor(np.ones((2, 6)))
bias = Tensor(np.ones((6,)))
(a + bias).sum().backward()
print(a.grad.shape, bias.grad.shape)                      # expect (2, 6) (6,)
```
