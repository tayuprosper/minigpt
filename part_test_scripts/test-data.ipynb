```python
from llm.tokenizer import CharTokenizer

text = open("data/fables.txt", encoding="utf-8").read()
tok = CharTokenizer(text)
print(tok.vocab_size)
print(tok.encode("Hello"))
print(tok.decode(tok.encode(text)) == text)
```

```python
from collections import Counter

tiny = "low lower lowest newer newest"

vocab = {i: ch for i, ch in enumerate(sorted(set(tiny)))}
ids = [{ ch : i for i, ch in vocab.items()}[ch] for ch in tiny]
print(len(ids), ids)

counts = Counter(zip(ids, ids[1:]))
for (a, b), n in counts.most_common(5):
    print(repr(vocab[a] + vocab[b]), n)
```

```python
from collections import Counter

after = Counter()

for i in range(len(text)-2):
    if text[i:i+2] == "th":
        after[text[i+2]] += 1

total = sum(after.values())
for ch, n in after.most_common(5):
    print(repr(ch), n, round(n / total, 3))
```

```python
import numpy as np

tok = CharTokenizer(text)
data = np.array(tok.encode(text))
V = tok.vocab_size


counts = np.zeros((V,V))
np.add.at(counts, (data[:-1], data[1:]), 1)
# print(counts)

probal  = (counts + 1) / (counts + 1).sum(axis=1, keepdims=True)
print(probal)
p_true = probal[data[:-1], data[1:]]
loss = -np.log(p_true).mean()
print(f"loss {loss: .3f} (knowing nothing: {np.log(V): .3f})")
```
